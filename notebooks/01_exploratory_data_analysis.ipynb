# Exploratory Data Analysis: Criteo CTR Prediction

This notebook explores the authentic 100,000-row sample of the **Criteo Display Advertising Challenge** dataset.

We inspect:
1. Schema and label distribution ($P(\text{click}=1)$).
2. Missing-value profiles across numerical ($I_1-I_{13}$) and categorical ($C_1-C_{26}$) features.
3. Right-skewness in numerical counts and the dampening effect of the $\log(1+x)$ transform.
4. Categorical cardinality and long-tail Zipfian frequency distributions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ctr_prediction.data import load_criteo_data
from ctr_prediction.validation import validate_schema

df = load_criteo_data()
report = validate_schema(df)
print(f"Total rows: {report['num_rows']:,}")
print(f"Positive CTR: {report['positive_rate']:.4%}")
```

```python
# Missing value rates
missing_rates = df.isna().mean().sort_values(ascending=False)
print("Top missing features:")
print(missing_rates.head(10))
```

```python
# Skewness comparison before and after log1p
skew_raw = df[['I1', 'I2', 'I3', 'I5']].skew()
skew_log = np.log1p(np.clip(df[['I1', 'I2', 'I3', 'I5']], 0, None)).skew()
print("Skewness Comparison:")
pd.DataFrame({'Raw Skew': skew_raw, 'Log1p Skew': skew_log})
```
